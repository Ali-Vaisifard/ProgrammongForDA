## File Handling

```python
DATA_DIR = Path("../data")
DATA_DIR.mkdir(exist.ok=True)
```

```python
# writing on Arden.txt in my DATA_DIR
Arden_file = DATA_DIR/ "Arden.txt"
with open(Arden_file, 'w') as file:
    file.write('Hello, my name is Arden \n It is Berlin campus \n We are studing Data Science.')
```

```python
# Opening file with read mode 
# read mode 'r', the file must already exist 
with open (Arden_file, 'r') as file:
    print(file.read())
    for line in file:
        print(file.read())
```

```python
# read fix number of characters
with open (Arden_file, 'r', encoding = "utf-8") as file:
    print(file.read(10))
```

```python
# Sequential reads from the current cursor position
with open (Arden_file, 'r') as file:
    con = file.read(4)
    con1 = file.read(10)
    con2 = file.read()

print(con) # read first chunk
print(con1) # read second chunk
print(con2) # read third chunk
```

```python
with open(Arden_file, 'a') as file:
    file.write('Hello, my name is Arden \n It is Berlin campus \n We are studing Data Science.')
```

```python
# Spliting line into words
with open(Arden_file, 'r') as file:
    for line in file:
        # split each line into words      
        words = line.split()
        for word in words:
            print(f'{word.lower()}\t1')
```

```python
# removing unwanted spaces with strip() lstrip() and rstrip()
e = "    Hello, my name is Arden     "
print("Original:", repr(e))
print("Left stripped:", repr(e.lstrip()))
print("Right stripped:", repr(e.rstrip()))
print("Both sides stripped:", repr(e.strip()))
```

## Basic numpy operations

```python
# arraay a and b
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
# add 1 to every elements
print(a+1)
print(b-2)
print(a.sum())
```

```python
# adding element wise a and b
c = np.add(a, b)
d = np.subtract(a, b)
print(c)
print(d)
```

```python
# multply a and b
Mul_ab = np.multiply(a, b)
Mul_ab
# np.divide() is for dividing
# np.power(a, 2) is for power of a
# np.mod(a, b) is for remainder
```

```python
# dot product a and b to get a scalar
dot_ab = np.vdot(a, b)
print(dot_ab)
```

```python
# dot product a and b to get a scalar
dot_ab1 = np.dot(a, b)
print(dot_ab1)
```

```python
# Quetient and reminer 
np.divmod(a,b)
```

```python
# logarityhm with base n
nplog = np.frompyfunc(log,2,1)
print(nplog(100,15))
```

```python
print(np.log2(a))
print(np.log10(b))
```
