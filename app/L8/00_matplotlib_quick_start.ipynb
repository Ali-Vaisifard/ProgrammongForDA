## Setup — import libraries and find the shared data folder

```python
%matplotlib inline

from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Find the shared data folder from the repository root or app/L8.
def data_path(filename):
    for folder in [Path.cwd(), *Path.cwd().parents]:
        file = folder / "data" / filename
        if file.is_file():
            return file
    raise FileNotFoundError(f"Cannot find {filename} from {Path.cwd()}")
```

## 1. One line plot

`plt.plot(x, y)` connects points. Always label the axes.

```python
months = [1, 2, 3, 4]
values = [4, 7, 6, 9]
plt.plot(months, values, marker="o")
plt.title("Example line plot")
plt.xlabel("Month")
plt.ylabel("Value")
plt.show()
```

## 2. Bar chart — compare categories

```python
plt.bar(["A", "B", "C"], [12, 8, 15])
plt.title("Example bar chart")
plt.show()
```

## 3. Scatter plot — compare two numbers

```python
plt.scatter([1, 2, 3, 4], [2, 4, 3, 7])
plt.title("Example scatter plot")
plt.show()
```

## 4. Histogram — show a distribution

```python
plt.hist([3, 4, 4, 5, 5, 5, 6, 7], bins=5)
plt.title("Example histogram")
plt.show()
```

## 5. Image — display an array of pixels

```python
import numpy as np
pixels = np.array([[0, 60, 120], [80, 160, 220], [120, 200, 255]])
plt.imshow(pixels, cmap="gray")
plt.colorbar(label="Pixel value")
plt.show()
```
