```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Works from the repository root OR from app/L9 in GitHub Codespaces.
def data_file(filename):
    for folder in [Path.cwd(), *Path.cwd().parents]:
        candidate = folder / "data" / filename
        if candidate.exists():
            return candidate
    raise FileNotFoundError(f"Cannot find data/{filename}. Open this notebook inside ProgrammingForDA.")
```

## Inspect the real dataset

Four flower measurements are features. The species is the target.

```python
flowers = pd.read_csv(data_file("iris_flowers.csv"))
print("Shape:", flowers.shape)
print(flowers.head(3))
print("Species:", flowers["species"].value_counts().to_dict())
```

## Create X and y

X contains the measurements; y contains the species names.

```python
X = flowers.drop(columns="species")
y = flowers["species"]
print("X:", X.shape, "y:", y.shape)
```

## Split, train and evaluate

Keep test observations separate from training observations.

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)
print("Test accuracy:", round(model.score(X_test, y_test), 3))
print("First 3 predictions:", model.predict(X_test.iloc[:3]).tolist())
```
